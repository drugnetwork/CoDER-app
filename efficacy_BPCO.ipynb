```python
%%R

library(dplyr)

disease_gene_path <- "data/bpco_gene_disease_associations.csv"
drug_gene_path <- "data/bpco_drug_gene_associations.csv"

disease_gene <- read.csv(disease_gene_path, stringsAsFactors = FALSE)
drug_gene <- read.csv(drug_gene_path, stringsAsFactors = FALSE)

disease_gene$geneSymbol <- toupper(trimws(disease_gene$geneSymbol))
drug_gene$gene_name <- toupper(trimws(drug_gene$gene_name))

disease_genes <- unique(na.omit(disease_gene$geneSymbol))

drug_gene_filtered <- drug_gene %>%
  filter(gene_name %in% disease_genes) %>%
  select(drug_name, gene_name) %>%
  distinct()

cat("BPCO disease-gene rows:", nrow(disease_gene), "\n")
cat("Unique BPCO disease genes:", length(disease_genes), "\n")
cat("BPCO drug-gene rows:", nrow(drug_gene_filtered), "\n")
cat("Unique BPCO drugs:", length(unique(drug_gene_filtered$drug_name)), "\n")

write.csv(disease_gene, "data/bpco_disease_genes_clean.csv", row.names = FALSE)
write.csv(drug_gene_filtered, "data/bpco_drug_gene_filtered_clean.csv", row.names = FALSE)
```

```python
%%R

library(dplyr)

disease_gene <- read.csv("data/bpco_gene_disease_associations.csv",
                         stringsAsFactors = FALSE)

drug_gene <- read.csv("data/bpco_drug_gene_associations.csv",
                      stringsAsFactors = FALSE)

disease_gene$geneSymbol <- toupper(trimws(disease_gene$geneSymbol))
drug_gene$gene_name <- toupper(trimws(drug_gene$gene_name))

bpco_genes <- unique(na.omit(disease_gene$geneSymbol))

drug_gene_filtered <- drug_gene %>%
  filter(gene_name %in% bpco_genes) %>%
  select(drug_name, gene_name) %>%
  distinct()

cat("BPCO disease-gene rows:", nrow(disease_gene), "\n")
cat("Unique BPCO disease genes:", length(bpco_genes), "\n")
cat("Filtered BPCO drug-gene rows:", nrow(drug_gene_filtered), "\n")
cat("Unique BPCO drugs:", length(unique(drug_gene_filtered$drug_name)), "\n")

write.csv(disease_gene,
          "data/bpco_disease_genes_clean.csv",
          row.names = FALSE)

write.csv(drug_gene_filtered,
          "data/bpco_drug_gene_filtered_clean.csv",
          row.names = FALSE)
```
