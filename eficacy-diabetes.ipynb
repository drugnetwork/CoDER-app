```python
import pandas as pd
import shutil
from pathlib import Path

repo = Path("Gene-Normalization-and-Mapping")
drug_dir = repo / "data" / "drug"
drug_dir.mkdir(parents=True, exist_ok=True)

# Copy diabetes disease-gene file
shutil.copy(
    "Gene-Normalization-and-Mapping/data/drug/diabetes_gene_disease_associations.csv",
    drug_dir / "diabetes_gene_disease_associations.csv"
)

# Convert diabetes drug-gene CSV to the Excel format expected by the R script
raw = pd.read_csv("/mnt/data/diabetes_drug_gene_associations.csv")

raw = raw[["drug_name", "gene_name"]].dropna()
raw["drug_name"] = raw["drug_name"].astype(str).str.strip()
raw["gene_name"] = raw["gene_name"].astype(str).str.strip()

sorted_df = (
    raw.groupby("drug_name")["gene_name"]
    .nunique()
    .reset_index(name="gene_count")
    .sort_values("gene_count", ascending=False)
)

output_excel = drug_dir / "diabetes_drug_gene_dataset.xlsx"

with pd.ExcelWriter(output_excel) as writer:
    sorted_df.to_excel(writer, sheet_name="Sorted", index=False, header=False)
    raw.to_excel(writer, sheet_name="Raw", index=False, header=False)

print("Saved:", output_excel)
print("Raw shape:", raw.shape)
print("Sorted shape:", sorted_df.shape)
print(sorted_df.head(10))
```
